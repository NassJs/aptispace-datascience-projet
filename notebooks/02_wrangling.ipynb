### 2. Chargement du dataset brut et Audit Initial

**À COMPLÉTER PAR L'ÉTUDIANT :**

### 3. Uniformisation des Formats de Dates

**À COMPLÉTER PAR L'ÉTUDIANT :**
Uniformisez la colonne temporelle pour la convertir dans un type datetime standardisé via votre module `src.data_clean`.

```python
# Le dataset ai_student_ ne contient pas de colonne temporelle :
# l'étape clean_dates n'est donc pas applicable.
df_clean = df_raw.copy()
print('Aucune colonne temporelle à convertir.')
df_clean.head()
```

### 4. Identification et Filtrage des Valeurs Aberrantes (Outliers)

**À COMPLÉTER PAR L'ÉTUDIANT :**
Identifiez les anomalies physiques et utilisez votre fonction `dc.handle_outliers` pour transformer ces valeurs aberrantes en NaNs.

```python
# Traitement des outliers sur les colonnes numériques clés :
# - GPA (Pre / Post) : intervalle plausible [0.0, 4.0]
# - Skill_Retention_Score : intervalle plausible [0.0, 100.0]
# - Weekly_GenAI_Hours : intervalle plausible [0.0, 60.0]
df_no_outliers = dc.handle_outliers(df_clean, ['Pre_Semester_GPA', 'Post_Semester_GPA'], 0.0, 4.0)
df_no_outliers = dc.handle_outliers(df_no_outliers, ['Skill_Retention_Score'], 0.0, 100.0)
df_no_outliers = dc.handle_outliers(df_no_outliers, ['Weekly_GenAI_Hours'], 0.0, 60.0)

df_no_outliers[['Pre_Semester_GPA', 'Post_Semester_GPA',
                'Skill_Retention_Score', 'Weekly_GenAI_Hours']].describe()
```

### 5. Imputation des valeurs manquantes

**À COMPLÉTER PAR L'ÉTUDIANT :**
Appliquez des stratégies d'imputation adaptées (interpolation temporelle, médiane, etc.) sur les valeurs manquantes générées ou initiales.

```python
# Imputation par médiane des NaN éventuellement créés par handle_outliers
cols_to_impute = ['Pre_Semester_GPA', 'Post_Semester_GPA',
                  'Skill_Retention_Score', 'Weekly_GenAI_Hours']
df_final = dc.impute_missing_values(df_no_outliers, cols_to_impute, 'median')

print('Valeurs manquantes restantes après imputation :')
print(df_final.isnull().sum().sum(), 'au total.')
```

### 6. Sauvegarde des données propres

Enregistrez vos données de base nettoyées dans le répertoire `data/processed/`.

```python
os.makedirs('../data/processed', exist_ok=True)
processed_path = '../data/processed/ai_student_cleaned.csv'
df_final.to_csv(processed_path, index=False)
print(f'Données propres sauvegardées dans : {processed_path}')
print(f'Dimensions finales : {df_final.shape}')
```
